```python
# customer analysis
```

```python
customer_bookings=pd.merge(bookings,cust,on="customer_id",how="left")
customer_bookings.head()
```

```python
# country wise average adr 
customer_bookings.groupby("country")["adr"].mean().sort_values(ascending=False)
```

```python
# booking status
import numpy as np
bookings["bookings_status"]=np.where(bookings["is_canceled"]==1,"Confirmed","Cancelled")
bookings.head()
```

```python
hotel_analysis = (
    bookings.groupby("hotel")
    .agg(
        total_bookings=("booking_id", "count"),
        average_adr=("adr", "mean"),
        
        cancellation_rate=("is_canceled", "mean")
    )
)

hotel_analysis
```

```python
customer_analysis=(
    bookings.groupby("customer_type")
    .agg(
        bookings=("booking_id","count"),
        average_adr=("adr","mean"),
        
    )
)
customer_analysis
```
